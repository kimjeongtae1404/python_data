```python
sentence = input("문장 입력>> ")
keywords = ["여행", "제주도", "맛집"]

for keyword in keywords:
    start = 0
    while True:
        index = sentence.find(keyword, start)
        if index == -1:
            break

        sentence = sentence[:index] + "#" + sentence[index:]
        start = index + len(keyword) + 1

print(sentence)
```

```python
exchange_rates={
    "USD":1200,
    "EUR":1400,
    "JPY":11,
}

print("---------환율 정보 추가/수정----------")
key=input("추가하거나 수정할 통화를 입력하세요: ")
value=int(input(f"{key}의 현재환율을 입력하세요: "))
exchange_rates[key]=value

print("---------환률계산기---------")
current=input(f"통화{list(exchange_rates.keys())}를 입력하세요: ")
price=int(input(f"변환할 금액({current})을 입력하세요: "))
print(f"환율 계산 결과:{round(price * exchange_rates[current], 2)} 원")
```

```python

categories = ["과학", "역사", "문학", "지리", "예술"]

answers = {
    "과학": "질소",
    "역사": "탑골공원",
    "문학": "햄릿",
    "지리": "나일강",
    "예술": "레오나르도 다빈치"
}

questions = {
    "과학": "지구의 대기 중 가장 많은 비율을 차지하는 기체는 무엇일까요?",
    "역사": "1919년 3월 1일, 민족대표 33인이 독립선언서를 낭독하고 만세 삼창을 불렀던 서울의 공원 이름은 무엇일까요?",
    "문학": "'사느냐 죽느냐, 그것이 문제로다'라는 유명한 대사가 나오는 셰익스피어의 희곡 제목은 무엇일까요?",
    "지리": "세계에서 가장 긴 강으로, 아프리카 북동부를 가로질러 지중해로 흘러 들어가는 이 강의 이름은 무엇일까요?",
    "예술": "'모나리자', '최후의 만찬' 등의 걸작을 남긴 르네상스 시대의 이탈리아 천재 화가는 누구일까요?"
}

hints = {
    "과학": "대기 중 약 78%를 차지하는 기체입니다.",
    "역사": "현재 서울 종로에 있는 공원입니다.",
    "문학": "덴마크 왕자 햄릿의 이야기를 다룬 비극입니다.",
    "지리": "이집트의 수도 카이로를 지나 지중해로 흘러갑니다.",
    "예술": "이탈리아 빈치에서 태어난 화가입니다."
}

completed_categories = []
score = 0

while len(completed_categories) < len(categories):
    selected_category = input(f"분야를 선택하세요 {categories}: ")

    if selected_category not in categories:
        print(f"분야를 다시 확인하세요: {categories}")
        continue
    if selected_category in completed_categories:
        print("이미 푼 분야입니다. 다른 분야를 선택하세요.")
        continue

    print(questions[selected_category])
    solved = False

    for attempt in range(2):
        answer = input("정답을 입력하세요: ")
        if answer == answers[selected_category]:
            print("정답입니다.")
            solved = True
            break
        if attempt < 2:
            print("틀렸습니다. 다시 시도하세요.")

    if solved== False:
        print("힌트:", hints[selected_category])
        answer = input("힌트를 보고 정답을 입력하세요: ")
        if answer == answers[selected_category]:
            print("정답입니다.")
            solved = True
        else:
            print("정답은", answers[selected_category], "입니다.")

    completed_categories.append(selected_category)
    if solved:
        score += 1

print("모든 분야를 풀었습니다!")
print("최종 점수:", score, "/", len(categories))
```
